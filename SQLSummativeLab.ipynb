### Step 2: Limited Edition California Product
The California sales rep team is interested in running promotional material for a new limited edition model they are releasing based on the famous San Francisco Cable Cars.
```python
# Replace None with your code
california_customers = pd.read_sql("""
SELECT customerNumber,
       customerName,
       contactFirstName,
       contactLastName,
       phone,
       city,
       state,
       CAST(creditLimit AS REAL) AS creditLimit
FROM customers
WHERE state = 'CA'
  AND country = 'USA'
  AND CAST(creditLimit AS REAL) > 25000
ORDER BY creditLimit DESC;
""", conn)

california_customers
```

### Step 4: USA Credit and Inventory Policy - Visual Required
The USA based product team is planning to adjust its credit policies and inventory allocation strategy based on the average credit limit of its customers. They would like to target this strategy at a state level with several goals in mind. 
1. Optimize inventory distribution:
    - States with higher average credit limits might be able to place larger orders, justifying priority in inventory allocation.
    - This could help ensure that states with more purchasing power always have products in stock.
2. Tailor credit policies:

    - Identify states where they might be too conservative or too liberal with credit limits.
3. Target marketing and sales efforts:
    - Focus promotional campaigns on states with higher credit limits, potentially leading to larger orders.

Execute a SQl query that returns the information required to address this ask.

```python
# Replace None with your code
# Step 4: Average credit limit by US state
# CAST creditLimit to a number, otherwise AVG would not work reliably on text values
avg_credit_by_state = pd.read_sql("""
SELECT state,
       ROUND(AVG(CAST(creditLimit AS REAL)), 2) AS avg_credit_limit,
       COUNT(*) AS number_of_customers
FROM customers
WHERE country = 'USA'
GROUP BY state
ORDER BY avg_credit_limit DESC;
""", conn)

avg_credit_by_state
```

Once you have the information returned in a dataframe, select an appropriate visualization to represent this data. You are welcome to utilize matplotlib, seaborn, or pandas plotting to produce your visual. Ensure that it has a verbose title and axis labels!

```python
# Replace None with your visual code
# Produce a visual to represent the average credit limit by state
plt.figure(figsize=(10, 6))
ax = sns.barplot(data=avg_credit_by_state, x='state', y='avg_credit_limit')

# Show the dollar value and the number of customers on top of each bar
for bar, n in zip(ax.patches, avg_credit_by_state['number_of_customers']):
    ax.annotate(f"${bar.get_height():,.0f}\n(n={n})",
                (bar.get_x() + bar.get_width() / 2, bar.get_height()),
                ha='center', va='bottom', fontsize=9)

plt.title('Average Customer Credit Limit by US State\n(n = number of customers in that state)', fontsize=14)
plt.xlabel('US State', fontsize=12)
plt.ylabel('Average Credit Limit (USD)', fontsize=12)
plt.ylim(0, avg_credit_by_state['avg_credit_limit'].max() * 1.2)
plt.tight_layout()
plt.show()
```

### Step 5: Top Customers - Visual Required
The company is approaching its 10 year anniversary and wants to acknowledge and thank its top customers with personalized communication. They have asked you to determine the top 10 customers based on the total amount of payments made, making sure to return the customer name for clarity. 

Execute a SQl query that returns the information required to address this ask.


```python
# Replace None with your code
# Step 5: Top 10 customers by total payments made
top_customers = pd.read_sql("""
SELECT c.customerNumber,
       c.customerName,
       SUM(CAST(p.amount AS REAL)) AS total_payments
FROM customers AS c
JOIN payments AS p
  ON c.customerNumber = p.customerNumber
GROUP BY c.customerNumber, c.customerName
ORDER BY total_payments DESC
LIMIT 10;
""", conn)

top_customers
```

### Step 6: Top Customer + Product Quantities
The product team is running an analysis on popular and common products sold to each customer in order to try and determine what new products they should be looking at to include in their catalog. This data will also be used by individual sales reps to recommend similar products to each customer next time they place an order. 

They have asked you to query information, for each customer, about any product they have purchased 10 or more units of. In addition they would like the full set of data to be sorted in ascending order by the total amount purchased.

Execute a SQl query that returns the information required to address this ask.

Hint: For this one, you'll need to make use of HAVING, GROUP BY, and ORDER BY — make sure you get the order of them correct!

```python
# Replace None with you code
# For each customer, every product they bought 10 or more units of
customer_product_quantities = pd.read_sql("""
SELECT c.customerNumber,
       c.customerName,
       p.productCode,
       p.productName,
       SUM(CAST(od.quantityOrdered AS INTEGER)) AS products_ordered
FROM customers AS c
JOIN orders AS o
  ON c.customerNumber = o.customerNumber
JOIN orderdetails AS od
  ON o.orderNumber = od.orderNumber
JOIN products AS p
  ON od.productCode = p.productCode
GROUP BY c.customerNumber, c.customerName, p.productCode, p.productName
HAVING SUM(CAST(od.quantityOrdered AS INTEGER)) >= 10
ORDER BY products_ordered ASC;
""", conn)

customer_product_quantities
```

The product team is looking into the demand across its different product lines. They are conducting a comprehensive review of its product portfolio and inventory management strategies. You have been asked to query data pertaining to each different product line, that contains the total quantity ordered and the total number of products for each respective product line. By examining the number of products and total quantity ordered for each product line, the company aims to:
1. Optimize product mix:
    - Identify which product lines have the most diverse offerings (high number of products)
    - Determine which lines are most popular (high total quantity ordered)
    - Compare if lines with more products necessarily lead to more orders
2. Improve inventory management:
    - Adjust stock levels based on the popularity of each product line
    - Identify potential overstocking in lines with low order quantities
    - Ensure adequate variety in high-performing product lines
3. Adjust marketing strategy:
    - Focus promotional efforts on product lines with high potential (many products but lower order quantities)
    - Capitalize on the popularity of high-performing lines in marketing campaigns
4. Advise Product development:
    - Invest in expanding product ranges for lines with high order quantities
    - Consider phasing out or revamping product lines with low numbers of products and low order quantities

Hint: Think about how you can and might have to utilize SQL DISTINCT statement

Execute a SQl query that returns the information required to address this ask.

```python
# Replace None with your code
product_line_analysis = pd.read_sql("""
SELECT pl.productLine,
       COUNT(DISTINCT p.productCode) AS number_of_products,
       SUM(CAST(od.quantityOrdered AS INTEGER)) AS total_quantity_ordered
FROM productlines AS pl
JOIN products AS p
  ON pl.productLine = p.productLine
JOIN orderdetails AS od
  ON p.productCode = od.productCode
GROUP BY pl.productLine
ORDER BY total_quantity_ordered DESC;
""", conn)

product_line_analysis
```

Once you have the information returned in a dataframe, select an appropriate visualization to represent the relationship between total quantity ordered and the number of products in order to perform a preliminary investigation into the question of if more products lead to more orders. You are welcome to utilize matplotlib, seaborn, or pandas plotting to produce your visual. Ensure that it has a verbose title and axis labels!

```python

# Visualize the relationship between product count and total quantity ordered
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=product_line_analysis,
    x='number_of_products',
    y='total_quantity_ordered',
    hue='productLine'
)

plt.title('Relationship Between Number of Products and Total Quantity Ordered by Product Line')
plt.xlabel('Number of Different Products in Each Product Line')
plt.ylabel('Total Quantity Ordered (Units)')

plt.legend(title='Product Line')
plt.tight_layout()
plt.show()
```

### Step 8: Remote Offices
Upper management is considering a shift to hybrid and remote work for certain locations and roles. They have tasked you with providing them data about employees who work in any office that has fewer than 5 total employees so they can better understand how to support those employees remotely when offices are shut down. 

Be sure to include information about the employees job and supervisor so management can adjust everyone to remote work properly.

Hint: Utilize a subquery to find the relevant offices

Execute a SQl query that returns the information required to address this ask.

```python
# Replace None with your code
# Employees who work in offices with fewer than 5 employees
# The subquery finds the offices; the main query returns the employees in those offices
# A self-join on employees gives us each employee's supervisor name
remote_office_employees = pd.read_sql("""
SELECT e.employeeNumber,
       e.firstName,
       e.lastName,
       e.jobTitle,
       e.officeCode,
       o.city AS office_city,
       e.reportsTo AS supervisorNumber,
       s.firstName || ' ' || s.lastName AS supervisorName
FROM employees AS e
JOIN offices AS o
  ON e.officeCode = o.officeCode
LEFT JOIN employees AS s
  ON e.reportsTo = s.employeeNumber
WHERE e.officeCode IN (
    SELECT officeCode
    FROM employees
    GROUP BY officeCode
    HAVING COUNT(*) < 5
)
ORDER BY e.officeCode, e.lastName;
""", conn)

remote_office_employees
```

## Reflection Question:

Describe how you decided on the subquery that you used in the query above? This answer can be technically in nature, describing your thought process in how the main query is utilizing the subquery to return the correct data.

## Type your answer in markdown here!

I divided the question into two parts: identifying offices with fewer than five employees and finding the employees who work in those offices.

The subquery groups employees by officeCode and uses HAVING COUNT(*) < 5 to identify offices with fewer than five employees. I used HAVING because the condition applies to an aggregate count.

The main query uses WHERE e.officeCode IN (...) to retrieve employees from those offices. I also joined the offices table to show each office's city and used a self-join with LEFT JOIN to display supervisors' names while retaining employees who have no supervisor.

```python
# Begin your code here
# Initial exploration of the IMDB movie database


# 1. Scope: how many titles, and which years are in the data?
pd.read_sql("""
SELECT start_year, COUNT(*) AS number_of_movies
FROM movie_basics
GROUP BY start_year
ORDER BY start_year;
""", conn4)
```

```python
# Count the null values in the 2010-2019 movies
data_quality = pd.read_sql("""
SELECT COUNT(*)                                              AS total_movies,
       SUM(CASE WHEN runtime_minutes IS NULL THEN 1 ELSE 0 END) AS missing_runtime,
       SUM(CASE WHEN genres IS NULL THEN 1 ELSE 0 END)          AS missing_genres,
       SUM(CASE WHEN r.movie_id IS NULL THEN 1 ELSE 0 END)      AS missing_rating
FROM movie_basics AS b
LEFT JOIN movie_ratings AS r
  ON b.movie_id = r.movie_id
WHERE b.start_year BETWEEN 2010 AND 2019;
""", conn4)
data_quality
```

```python
# 3. Pull the clean working set: 2010-2019 movies that have no null values
movies = pd.read_sql("""
SELECT b.movie_id,
       b.primary_title,
       b.start_year,
       b.runtime_minutes,
       b.genres,
       r.averagerating,
       r.numvotes
FROM movie_basics AS b
JOIN movie_ratings AS r
  ON b.movie_id = r.movie_id
WHERE b.start_year BETWEEN 2010 AND 2019
  AND b.genres IS NOT NULL
  AND b.runtime_minutes IS NOT NULL;
""", conn4)

print(len(movies), "movies in the working set")
movies.head()
```

The `genres` column stores several genres in one text value (for example `Action,Crime,Drama`). To analyze one genre at a time I split that text and give each genre its own row. A movie with three genres therefore counts once in each of its three genres.

```python
# Split the multi-genre text so each genre gets its own row, then summarize by genre
movies_by_genre = movies.assign(genre=movies['genres'].str.split(',')).explode('genre')

genre_summary = (movies_by_genre
                 .groupby('genre')
                 .agg(number_of_movies=('movie_id', 'count'),
                      avg_rating=('averagerating', 'mean'),
                      median_votes=('numvotes', 'median'),
                      avg_runtime=('runtime_minutes', 'mean'))
                 .round(2)
                 .reset_index())

# Keep genres with at least 500 movies so that the averages are reliable
genre_summary = (genre_summary[genre_summary['number_of_movies'] >= 500]
                 .sort_values('avg_rating', ascending=False)
                 .reset_index(drop=True))
genre_summary
```

```python
#  Visual: average rating by genre, with the median number of votes shown on each bar
plt.figure(figsize=(10, 6))

ax = sns.barplot(data=genre_summary, y='genre', x='avg_rating')

for i, row in genre_summary.iterrows():
    ax.text(row['avg_rating'] + 0.05, i,
            f"({row['median_votes']:,.0f} votes)",
            va='center', fontsize=9)

plt.title('Average IMDb Rating by Genre (2010–2019)')
plt.xlabel('Average Rating')
plt.ylabel('Genre')
plt.xlim(0, 10)
plt.show()
```

```python
#  Does a longer runtime go with a higher rating?
# Ignore extreme runtimes (under 40 min or over 300 min)
normal_runtime = movies[(movies['runtime_minutes'] >= 40) & (movies['runtime_minutes'] <= 300)]
print("Correlation between runtime and rating:",
      round(normal_runtime['runtime_minutes'].corr(normal_runtime['averagerating']), 3))
```
